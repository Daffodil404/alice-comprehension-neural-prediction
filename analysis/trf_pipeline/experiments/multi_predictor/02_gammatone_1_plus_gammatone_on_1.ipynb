# Gammatone-1 + gammatone-on-1 all-subject TRF experiment

Run a composed acoustic model containing the 1-band gammatone predictor and the 1-band acoustic-onset predictor for every subject discovered by the current Eelbrain pipeline. This model tests whether broadband onset information improves held-out EEG prediction beyond the 1-band gammatone representation. Shared TRF settings, result extraction, QC, and checkpoint writing live in `../_shared.py`.

```python
EXPERIMENT_ID = 'gammatone-1-plus-gammatone-on-1'
PREDICTOR = 'gammatone-1 + gammatone-on-1'
```
