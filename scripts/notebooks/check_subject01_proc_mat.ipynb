# Check proc.zip


```python
from io import BytesIO
from pathlib import Path
from pprint import pprint
from zipfile import ZipFile
from scipy.io import loadmat
```

```python
PROC_ZIP = Path("/Users/yanyuwoo/Data/r/proc.zip")
SUBJECT = "S01"

with ZipFile(PROC_ZIP) as archive:
    mat_bytes = archive.read(
        f"timelock-preprocessing/{SUBJECT}.mat"
    )

mat = loadmat(
    BytesIO(mat_bytes),
    simplify_cells=True,
)

print(mat.keys())
```

```python
proc = mat["proc"]
print(proc.keys())
```

```python
print("Subject:", proc["subject"])
print("Implicit reference:", proc["implicitref"])
print("Reference channels:", proc["refchannels"])
print("Bad channels:", proc["rejections"]["badchans"])
```
